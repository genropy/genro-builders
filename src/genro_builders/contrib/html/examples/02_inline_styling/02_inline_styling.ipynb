## 1. CSS as kwargs

Any kwarg whose name matches a *curated CSS root* (or starts
with a root followed by `_`) is routed into the `style`
attribute. The renderer converts underscores to dashes:
`background_color` → `background-color`.

Roots include: `color`, `background`, `font`, `text`, `border`,
`padding`, `margin`, `width`, `height`, `top`, `left`, `right`,
`bottom`, `display`, `visibility`, `opacity`, `overflow`,
`position`, `cursor`, `z_index`, `flex`, `gap`,
`align_content`, `justify_content`, `align_items`,
`justify_items`, `line_height`, `white_space`,
`vertical_align`, `resize`, `float`, `clear`.

```python
from IPython.display import HTML

from genro_builders.contrib.html import HtmlBuilderHandler


class Section1(HtmlBuilderHandler):
    def main(self, root):
        body = root.body()
        body.div(
            "Card",
            color="white", background="#3498db", padding="12px",
        )


p1 = Section1(); p1.create(); p1.build()
print(p1.render(pretty=True))
```

```python
HTML(p1.render())
```

## 2. Sub-property kwargs

A kwarg that *starts* with a root followed by `_` is also
routed into style. `padding_top` matches root `padding` →
`padding-top`. Same for `font_size`, `text_align`, `border_radius`,
`margin_left`, and so on.

```python
class Section2(HtmlBuilderHandler):
    def main(self, root):
        body = root.body()
        body.div(
            "Stylish heading",
            color="white", background="#2c3e50",
            padding_top=12, padding_bottom=12, padding_left=24,
            font_size="14px", text_align="center",
        )


p2 = Section2(); p2.create(); p2.build()
print(p2.render(pretty=True))
```

```python
HTML(p2.render())
```

## 3. Explicit `style=` merged with kwargs

When both an explicit `style="..."` string and CSS kwargs are
passed, the renderer parses the string into a dictionary and
then applies the kwargs over it: **kwargs win on collision**.
The modern specific syntax overrides the legacy catch-all.

Below: the explicit `style` brings `padding: 6px`; the kwarg
`color="white"` overrides the explicit `color: blue`.

```python
class Section3(HtmlBuilderHandler):
    def main(self, root):
        body = root.body()
        body.div(
            "Merged styling",
            style="color: blue; padding: 6px",
            color="white", background="#e67e22",
        )


p3 = Section3(); p3.create(); p3.build()
print(p3.render(pretty=True))
```

```python
HTML(p3.render())
```

## 4. `style_<prop>` escape

For CSS properties outside the curated root list (for example
`aspect-ratio`, `accent-color`, modern properties), the prefix
`style_` forces the kwarg into the style attribute, stripping
the prefix: `style_aspect_ratio="3 / 1"` → `aspect-ratio: 3 / 1`.

```python
class Section4(HtmlBuilderHandler):
    def main(self, root):
        body = root.body()
        body.div(
            "Aspect ratio 3:1",
            background="#16a085", color="white",
            style_aspect_ratio="3 / 1",
            display="flex",
            align_items="center",
            justify_content="center",
        )


p4 = Section4(); p4.create(); p4.build()
print(p4.render(pretty=True))
```

```python
HTML(p4.render())
```

## 5. `rounded` macro

Genro macro: `rounded=N` sets all four `border-*-radius` to
`Npx`. Single-underscore sub-kwargs restrict the corners
(matching the legacy convention):
`rounded_top=N` (top corners only), `rounded_bottom`,
`rounded_left`, `rounded_right`, or per-corner names
(`rounded_top_left`, `rounded_top_right`,
`rounded_bottom_left`, `rounded_bottom_right`).

Macro names are chosen so they never collide with CSS roots,
so the macro lookup wins with no ambiguity. Numeric values
are emitted with `px`; strings pass through
(`rounded="50%"` for circles).

```python
class Section5(HtmlBuilderHandler):
    def main(self, root):
        body = root.body()
        body.div(
            "All corners (rounded=12)",
            background="#9b59b6", color="white", padding="12px",
            rounded=12,
        )
        body.div(
            "Top corners only (rounded_top=20)",
            background="#8e44ad", color="white", padding="12px",
            rounded_top=20,
        )


p5 = Section5(); p5.create(); p5.build()
print(p5.render(pretty=True))
```

```python
HTML(p5.render())
```

## 6. `transform` macro

Compose `transform: <fn> <fn> ...` from single-underscore
sub-kwargs. Recognised: `transform_rotate=deg`,
`transform_scale=factor`, `transform_translate="<value>"`,
`transform_translate_x=px`, `transform_translate_y=px`,
`transform_skew_x=deg`, `transform_skew_y=deg`. Any other
sub-kwarg becomes `<name>(value)`.

```python
class Section6(HtmlBuilderHandler):
    def main(self, root):
        body = root.body()
        body.div(
            "Rotate + scale",
            background="#f1c40f", color="#2c3e50", padding="12px",
            transform_rotate=-3, transform_scale=0.95,
            display="inline-block",
        )


p6 = Section6(); p6.create(); p6.build()
print(p6.render(pretty=True))
```

```python
HTML(p6.render())
```

## 7. `filter` macro

Compose `filter: <fn> <fn> ...` from single-underscore
sub-kwargs. Recognised: `filter_blur=px`, `filter_contrast=N`,
`filter_brightness=N`, `filter_saturate=N`, `filter_grayscale=N`,
`filter_sepia=N`, `filter_opacity=N`, `filter_invert=N`,
`filter_rotate=deg` (emits `hue-rotate`),
`filter_drop_shadow="<value>"`.

```python
class Section7(HtmlBuilderHandler):
    def main(self, root):
        body = root.body()
        body.div(
            "Blurred and reduced contrast",
            background="#e74c3c", color="white", padding="12px",
            filter_blur=2, filter_contrast=80,
        )


p7 = Section7(); p7.create(); p7.build()
print(p7.render(pretty=True))
```

```python
HTML(p7.render())
```
