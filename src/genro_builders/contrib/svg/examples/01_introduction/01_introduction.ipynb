## 1. Hello SVG

An image is a subclass of `SvgBuilderHandler` that implements
`main(self, root)`. Start from a root `<svg>` element with a
`viewBox`, then place shapes inside it.

The lifecycle mirrors the HTML dialect: `create()` populates
the source, `build()` materialises it, `render()` serialises
to SVG. Void tags are emitted self-close XHTML-style with a
space before the slash (`<rect />`), following the W3C SVG
convention.

```python
from IPython.display import SVG

from genro_builders.contrib.svg import SvgBuilderHandler


class HelloSvg(SvgBuilderHandler):
    def main(self, root):
        svg = root.svg(viewBox="0 0 100 100", width=200, height=200)
        svg.rect(x=10, y=10, width=80, height=80, fill="#3498db")
        svg.circle(cx=50, cy=50, r=30, fill="#e74c3c")


page = HelloSvg()
page.create()
page.build()
print(page.render())
```

```python
SVG(page.render())
```

## 2. Lifecycle: source vs built

Same model as the HTML dialect. `source` holds the user recipe,
`built` is the materialised tree the renderer walks over. For
elementary shapes they coincide; the distinction matters once
components, iterate, or pointers come into play.

```python
print("--- source ---")
print(page.source.to_xml())
print("--- built ---")
print(page.built.to_xml())
```

## 3. Kebab-case attributes

SVG presentation attributes (`stroke-width`, `fill-opacity`,
`font-size`, ...) use hyphens, but Python identifiers cannot
contain them. Write underscores (`stroke_width`) and the
renderer converts them to `stroke-width` automatically.
The full list lives in `_KEBAB_ATTRS` in `svg_builder.py`.

```python
class StyledSvg(SvgBuilderHandler):
    def main(self, root):
        svg = root.svg(viewBox="0 0 100 100", width=200, height=200)
        svg.circle(
            cx=50, cy=50, r=40,
            fill="none",
            stroke="#2c3e50",
            stroke_width=4,
            stroke_dasharray="6 3",
        )


styled = StyledSvg()
styled.create()
styled.build()
print(styled.render())
```

```python
SVG(styled.render())
```

## 4. Nested groups

The `<g>` (group) element is a container that applies a
transform to its children. Use it to position blocks of
shapes without recomputing coordinates by hand.

```python
class GroupedSvg(SvgBuilderHandler):
    def main(self, root):
        svg = root.svg(viewBox="0 0 200 100", width=400, height=200)
        left = svg.g(transform="translate(10,10)")
        left.rect(x=0, y=0, width=60, height=60, fill="#3498db")
        right = svg.g(transform="translate(110,10)")
        right.rect(x=0, y=0, width=60, height=60, fill="#e74c3c")


grouped = GroupedSvg()
grouped.create()
grouped.build()
print(grouped.render())
```

```python
SVG(grouped.render())
```

## 5. Text inside SVG

The `<text>` element accepts a string value as its content.
Typographic attributes follow the kebab-case convention
(`text_anchor`, `font_size`, `font_family`).

```python
class LabeledSvg(SvgBuilderHandler):
    def main(self, root):
        svg = root.svg(viewBox="0 0 200 60", width=400, height=120)
        svg.rect(x=0, y=0, width=200, height=60, fill="#ecf0f1")
        svg.text(
            "Hello SVG",
            x=100, y=40,
            text_anchor="middle",
            font_size=24,
            font_family="sans-serif",
            fill="#2c3e50",
        )


labeled = LabeledSvg()
labeled.create()
labeled.build()
print(labeled.render())
```

```python
SVG(labeled.render())
```
