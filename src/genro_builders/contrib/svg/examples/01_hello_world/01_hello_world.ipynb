# 01 — Hello World (SVG)

The simplest possible example with `SvgBuilderHandler`.

**What you learn here:**
- Subclass `SvgBuilderHandler` and bind the SVG dialect to the handler.
- Populate the `source` Bag inside `main(self, root)`.
- The three-phase lifecycle `create()` → `build()` → `render()`.
- Kebab-case attribute names in SVG: `stroke_width` (Python) → `stroke-width` (SVG).
- Display the SVG inline in the notebook.

## 1. Define an SVG image

Start from a root `<svg>` element with a `viewBox`; place shapes inside it.

```python
from genro_builders.contrib.svg import SvgBuilderHandler


class HelloSvg(SvgBuilderHandler):
    def main(self, root):
        svg = root.svg(viewBox="0 0 100 100", width=200, height=200)
        svg.rect(x=10, y=10, width=80, height=80, fill="#3498db")
        svg.circle(cx=50, cy=50, r=30, fill="#e74c3c")
```

## 2. `create()` → `build()` → `render()`

Same lifecycle as the HTML dialect. The SVG dialect always emits XML well-formed output, with void tags written as `<rect />` (space before the slash, following the W3C XHTML/SVG convention).

```python
page = HelloSvg()
page.create()
page.build()

svg_text = page.render()
print(svg_text)
```

## 3. Inline preview (Jupyter)

```python
from IPython.display import SVG

SVG(svg_text)
```

## 4. Kebab-case attributes

SVG presentation attributes (`stroke-width`, `fill-opacity`, `font-size`, ...) use hyphens, but Python identifiers cannot contain them.
Write underscores (`stroke_width`) and the renderer converts them to `stroke-width` automatically.
The full list lives in `_KEBAB_ATTRS` in `svg_builder.py`.

```python
class StyledSvg(SvgBuilderHandler):
    def main(self, root):
        svg = root.svg(viewBox="0 0 100 100", width=200, height=200)
        svg.circle(
            cx=50, cy=50, r=40,
            fill="none",
            stroke="#2c3e50",
            stroke_width=4,
            stroke_dasharray="6 3",
        )


styled = StyledSvg()
styled.create()
styled.build()
print(styled.render())
```

```python
SVG(styled.render())
```

## 5. Nested groups

The `<g>` (group) element is a container that applies transforms to its children.

```python
class GroupedSvg(SvgBuilderHandler):
    def main(self, root):
        svg = root.svg(viewBox="0 0 200 100", width=400, height=200)
        left = svg.g(transform="translate(10,10)")
        left.rect(x=0, y=0, width=60, height=60, fill="#3498db")
        right = svg.g(transform="translate(110,10)")
        right.rect(x=0, y=0, width=60, height=60, fill="#e74c3c")


grouped = GroupedSvg()
grouped.create()
grouped.build()
print(grouped.render())
```

```python
SVG(grouped.render())
```

## 6. Text inside SVG

The `<text>` element accepts a string value as the node content.

```python
class LabeledSvg(SvgBuilderHandler):
    def main(self, root):
        svg = root.svg(viewBox="0 0 200 60", width=400, height=120)
        svg.rect(x=0, y=0, width=200, height=60, fill="#ecf0f1")
        svg.text(
            "Hello SVG",
            x=100, y=40,
            text_anchor="middle",
            font_size=24,
            font_family="sans-serif",
            fill="#2c3e50",
        )


labeled = LabeledSvg()
labeled.create()
labeled.build()
print(labeled.render())
```

```python
SVG(labeled.render())
```
