# 01 — Hello World (SVG)

Il più semplice esempio possibile con `SvgBuilderHandler`.

**Cosa si impara qui:**
- Sottoclassare `SvgBuilderHandler` e bindare il dialetto SVG al handler.
- Popolare la `source` Bag dentro `main(self, root)`.
- Il ciclo a tre fasi `create()` → `build()` → `render()`.
- Attributi kebab-case in SVG: `stroke_width` (Python) → `stroke-width` (SVG).
- Visualizzare l'SVG inline nel notebook.

## 1. Definire un'immagine SVG

Si parte sempre da un elemento radice `<svg>` con `viewBox`; dentro si inseriscono le forme.

```python
from genro_builders.contrib.svg import SvgBuilderHandler


class HelloSvg(SvgBuilderHandler):
    def main(self, root):
        svg = root.svg(viewBox="0 0 100 100", width=200, height=200)
        svg.rect(x=10, y=10, width=80, height=80, fill="#3498db")
        svg.circle(cx=50, cy=50, r=30, fill="#e74c3c")
```

## 2. `create()` → `build()` → `render()`

Stesso ciclo del dialetto HTML. Il dialetto SVG produce sempre XML well-formed con void tag `<rect />` (spazio prima dello slash, come da convenzione W3C XHTML/SVG).

```python
page = HelloSvg()
page.create()
page.build()

svg_text = page.render()
print(svg_text)
```

## 3. Preview inline (Jupyter)

```python
from IPython.display import SVG

SVG(svg_text)
```

## 4. Attributi kebab-case

Gli attributi di presentazione SVG (`stroke-width`, `fill-opacity`, `font-size`, ...) usano i trattini, ma Python non li accetta negli identificatori.
Si scrive l'underscore (`stroke_width`) e il renderer lo converte automaticamente in `stroke-width`.
L'elenco completo è in `_KEBAB_ATTRS` dentro `svg_builder.py`.

```python
class StyledSvg(SvgBuilderHandler):
    def main(self, root):
        svg = root.svg(viewBox="0 0 100 100", width=200, height=200)
        svg.circle(
            cx=50, cy=50, r=40,
            fill="none",
            stroke="#2c3e50",
            stroke_width=4,
            stroke_dasharray="6 3",
        )


styled = StyledSvg()
styled.create()
styled.build()
print(styled.render())
```

```python
SVG(styled.render())
```

## 5. Gruppi annidati

L'elemento `<g>` (group) è un container che applica trasformazioni ai figli.

```python
class GroupedSvg(SvgBuilderHandler):
    def main(self, root):
        svg = root.svg(viewBox="0 0 200 100", width=400, height=200)
        left = svg.g(transform="translate(10,10)")
        left.rect(x=0, y=0, width=60, height=60, fill="#3498db")
        right = svg.g(transform="translate(110,10)")
        right.rect(x=0, y=0, width=60, height=60, fill="#e74c3c")


grouped = GroupedSvg()
grouped.create()
grouped.build()
print(grouped.render())
```

```python
SVG(grouped.render())
```

## 6. Testo dentro SVG

L'elemento `<text>` accetta un valore stringa come contenuto del nodo.

```python
class LabeledSvg(SvgBuilderHandler):
    def main(self, root):
        svg = root.svg(viewBox="0 0 200 60", width=400, height=120)
        svg.rect(x=0, y=0, width=200, height=60, fill="#ecf0f1")
        svg.text(
            "Hello SVG",
            x=100, y=40,
            text_anchor="middle",
            font_size=24,
            font_family="sans-serif",
            fill="#2c3e50",
        )


labeled = LabeledSvg()
labeled.create()
labeled.build()
print(labeled.render())
```

```python
SVG(labeled.render())
```
